## 3. Criação de DataFrames

Aqui, experimentei criar um DataFrame a partir de uma lista Python.

O DataFrame é uma estrutura de dados organizada em linhas e colunas, semelhante a uma tabela, que permite trabalhar com dados estruturados no Spark.

```python
dados = [
    ("Ana", "Dados", 25),
    ("João", "Engenharia", 30),
    ("Maria", "Dados", 28),
    ("Pedro", "Marketing", 35)
]

df = spark.createDataFrame(
    dados,
    ["nome", "departamento", "idade"]
)

df.show()
```

**Referência:** [DataFrame — Documentação oficial](https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/dataframe.html)

## 4. Operações comuns com DataFrames

Nesta seção, reuni algumas operações utilizadas para explorar e manipular DataFrames.

### `show()`

Utilizada para visualizar os registros de um DataFrame.

### `printSchema()`

Permite observar a estrutura e os tipos de dados das colunas.

### `select()`

Utilizada para selecionar colunas específicas.

### `filter()`

Permite filtrar registros a partir de condições.

### `withColumn()`

Utilizada para criar ou modificar colunas.

```python
# Visualizar os registros
df.show()

# Explorar a estrutura
df.printSchema()

# Selecionar colunas
df.select("nome", "idade").show()

# Filtrar registros
df.filter(df["idade"] > 27).show()

# Criar uma nova coluna
df.withColumn(
    "idade_futura",
    df["idade"] + 1
).show()
```

## 5. Agrupamentos e agregações

Nesta etapa, explorei operações utilizadas para agrupar registros e calcular métricas.

### `groupBy()`

Agrupa registros com base em uma ou mais colunas.

### `count()`

Contabiliza registros.

### `agg()`

Permite realizar diferentes operações de agregação em um mesmo agrupamento.

### `avg()`, `sum()`, `min()` e `max()`

Funções utilizadas para calcular média, soma, mínimo e máximo.

```python
from pyspark.sql import functions as F

# Contagem de registros por departamento
df.groupBy("departamento").count().show()

# Agregações
df.groupBy("departamento").agg(
    F.avg("idade").alias("idade_media"),
    F.max("idade").alias("idade_maxima")
).show()
```

**Referência:** [Functions — Documentação oficial](https://spark.apache.org/docs/latest/api/python/reference/pyspark.sql/functions.html)
